```r
#FAMILIES OF DISTRIBUTIONS 

#runif(n, size)       : random uniform  experiment
#rbinom(n, size, p)   : random binomial experiment
#rpois(n, lambda)     : random poisson experiment
#rexp(n, lambda)      : random exponential experiment
#rnorm(n, mean, variance)  : random normal experiment
#  -> pnorm for PDF, -> dnorm for cdf are very useful
#   indeed pnorm(z)=1/square(2Pi) INT[ e^{z^2/2} dz]

#if not r, but d: pdf
#if not r, but p: CDF
#if not r, but q: nth percentile
```

```r
sample(1:20, 5, replace=FALSE)    #simple sampling with no repetitions
```

```r
unif_exp <- runif(100, min=0, max=5)   #random uniform experiment
                                       #outputs a vector of values between min and max, same probability of being chosen

length(unif_exp)

unif_exp

head(unif_exp, 10)    #first 10 values of the list
tail(unif_exp, 10)    #last 10 entries of the list

mean(unif_exp)

plot(seq(1,100,by=1.0), unif_exp, xlab="number of experiment", ylab="outcome", main="outcomes of simulations")
```

```r
bin_exp <- rbinom(100, 20, 0.7)    #random binomial experiment
                                   #outputs a vector of successes over n simulations

                                   #n=100 simulations
                                   #for each simulation, size=20 Bernoulli trials: 0,1,0,0,0,1,1,0
                                   #prob=0.7 the probability of a success, i.e. 1, for a single Bernoulli trial

length(bin_exp)

bin_exp

mean(bin_exp)

sd(bin_exp)

plot(seq(1.0,100,by=1.0), bin_exp, type="l", xlab="number of experiment", ylab="successes", main="plot of successes for 100 binomial experiments")

y <- seq(1,100, by=1.0)
plot(y, dbinom(y, 20, 0.7), type="b", main="pdf of Binom(20,0.7)")
```

```r
ppois(3, 10)    #CDF for a poisson process F_{lambda=10}(x=3)=P(X<=3)

dpois(3, 10)    #pmf/pdf for a poisson process f_{lambda=10}(x=3)=P(X=3)

poi_exp <- rpois(100, 10)     #random poisson experiment, when at a time event occurs 10 times
                              #output a vector of occuring events in 100 simulations on time intervals of same length          
plot(seq(1,100,by=1.0), poi_exp, type="l", main="events occurring in same time interval", xlab="simulation", ylab="number of events")  


y <- seq(1, 100, by=1.0)            #plot pmf
plot(y, dpois(y, 10), type="b", main="probability density function for Poi(10)", xlab="x", ylab="pdf")
```

```r
rnorm(100, mean=0, sd=1)

y <- seq(-50, 50, by=0.5)
plot(y, dnorm(y, mean=0, sd=1), type="l", main="pdf of standard normal distribution", xlab="x", ylab="pdf")

pnorm(1)-pnorm(-1) #probability of being in [-sd,+sd]      #The 68-95-99.7% empirical rule
pnorm(2)-pnorm(-2) #probability of being in [-2sd,+2sd]
pnorm(3)-pnorm(-3) #probability of being in [-3sd,+3sd]

qnorm(0.9)    #90th percentile
```

```r
rexp(100, 3)    #random exponential experiment
                #100 simulations of occurring events, considering for each jump there are 3 events
```
