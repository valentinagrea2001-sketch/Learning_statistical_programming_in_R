```r
#FIND THE CONFIDENCE INTERVAL

#study on axorexia:
#cognitive behavioural therapy used to treat a sample of anorexic teenage girls: cb
#control group of anorexic girls not receiving cb: c
#not considered: f

#RMK: girls are randomly assigned to experimental and control groups

anor <- read.table("https://stat4ds.rwth-aachen.de/data/Anorexia.dat", header=TRUE)
anor
```

```r
#We want to compare values before and after some time in cb

anor_diff_cb <- anor$after[anor$therapy=="cb"]-anor$before[anor$therapy=="cb"]   #experimental group

anor_diff_c <- anor$after[anor$therapy=="c"]-anor$before[anor$therapy=="c"]   #control group

summary(anor_diff_cb)

summary(anor_diff_c)
```

```r
boxplot(anor_diff_cb, anor_diff_c, names=c("experimental group", "control group"), col=c("lightblue","lightgreen"))

#we can immediately notice that for those girls receiving cb, the mean of improvement is higher
#and that less girls have done worse

#while in the control group, still the majority of girls scored worst results, i.e. more weight loss
```

```r
#Find a 95% confidence interval for the mean of the experimental group
#what is the average improvement in terms of weight for teenage anorexic girls?

alpha <- 0.05

mu <- mean(anor_diff_cb)  #sample mean

se <- sqrt(var(anor_diff_cb)/length(anor_diff_cb))   #sample standard erro

t <- qt(1 - alpha/2, df = length(anor_diff_cb) - 1)  #t-value from t distribution

print("Confidence interval for the sample mean of weigths improvement in experimental group")

mu - t * se

mu + t * se
```

```r
#now find the confidence interval for the control group

muc <- mean(anor_diff_c)

sec <- sqrt(var(anor_diff_c)/length(anor_diff_c))

tc <- qt(1 - alpha/2, df=length(anor_diff_c)-1)

print("Confidence interval in the control group")

muc - tc * sec

muc + tc * sec
```

```r
#CONCLUSIONS:
#The average of weights improvement is substantially better
#for girls who received cb treatment

#Control group girls weight became even worse
```
