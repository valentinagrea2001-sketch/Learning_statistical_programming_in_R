```r
#FIND THE CONFIDENCE INTERVAL

#study on axorexia:
#cognitive behavioural therapy used to treat a sample of anorexic teenage girls: cb
#control group of anorexic girls not receiving cb: c
#not considered: f

#RMK: girls are randomly assigned to experimental and control groups

anor <- read.table("https://stat4ds.rwth-aachen.de/data/Anorexia.dat", header=TRUE)
anor
```

```r
#We want to compare values before and after some time in cb

anor_diff_cb <- anor$after[anor$therapy="cb"]-anor$before[anor$therapy="cb"]   #experimental group

anor_diff_c <- anor$after[anor$therapy="c"]-anor$before[anor$therapy="c"]   #control group
```
