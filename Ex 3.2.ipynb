```r
#SAMPLING DISTRIBUTION OF STATISTICS

#If statistics= sample mean
#->-> by the CLT, the sampling distribution is approximatively normal

#Try for other statistics if the CLT holds:
#sample mean: OK
#sample proportion: OK
#sample variance: NO
#sample standard deviation: NO
#sample median: OK but only under certain conditions
#sample sum: OK
#sample min/max: NO

install.packages("moments")    #to import skewness

# Load the library
library(moments)
```

```r
#We choose this time an exponential random experiment
#a large population of N=10000, lambda=5

exp_experiment <- rexp(10000,5)

nsamples <- 5000

s <- list()
for (i in 1:nsamples) {
    s[[i]] <- sample(exp_experiment, 300)   #sample sets n=500
}
```

```r
#statistics= proportion > 0.2 ?

prop <- c()                   #to store proportions
for (i in 1:nsamples) {
    b <- s[[i]] > 0.2         #returns a boolean vector of 1,0
    prop <- c(prop, mean(b))  #adds the proportion
}

hist(prop, main="sampling distribution of proportion > 0.2", xlab="values", ylab="proportions")
```

```r
#statistics = variance

v <- c()
for (i in 1:nsamples) {
    v <- c(v, var(s[[i]]))
}

mean(v)

median(v)

skewness(v)

hist(v, main="sampling distribution of variance")       #skewn to the left
```

```r
#statistics = standard deviation

standard <- c()
for (i in 1:nsamples){
    standard <- c(standard, sd(s[[i]]))
}

mean(standard) 

median(standard)

skewness(standard)   

hist(standard, main="sampling distribution of standard deviation")    #skewn to the left
```

```r
#statistics = median

med <- c()
for (i in 1:nsamples){
    med <- c(med, median(s[[i]]))
}

mean(med)

median(med)

hist(med, main="sampling distribution of median")
```

```r
#statistcs = sum

su <- c()
for (i in 1:nsamples) {
    su <- c(su, sum(s[[i]]))
}

hist(su, main="sampling distribution of sum")
```

```r
#statistics = min

m <- c()
for (i in 1:nsamples){
    m <- c(m, min(s[[i]]))
}

hist(m, main="Sampling distribution of min")
```

```r
#statistics = max

ma <- c()
for (i in 1:nsamples){
    ma <- c(ma, max(s[[i]]))
}

hist(ma, main="Sampling distribution of max")
```
