```r
#SAMPLING DISTRIBUTION OF STATISTICS

#If statistics= sample mean
#->-> by the CLT, the sampling distribution is approximatively normal

#Try for other statistics if the CLT holds:
#sample proportion: OK
#sample variance: NO
#sample standard deviation: OK
#sample median: NO
#sample sum: OK
#sample min/max: NO

```

```r
#We choose this time an exponential random experiment
#a large population of N=10000, lambda=5

exp_experiment <- rexp(10000,5)

s <- list()
for (i in 1:1000) {
    s[[i]] <- sample(exp_experiment, 500)   #sample sets n=500
}
```

```r
#statistics= proportion > 0.2 ?

prop <- c()                   #to store proportions
for (i in 1:1000) {
    b <- s[[i]] > 0.2         #returns a boolean vector of 1,0
    prop <- c(prop, mean(b))  #adds the proportion
}

hist(prop, main="sampling distribution of proportion > 0.2", xlab="values", ylab="proportions")
```

```r
#statistics = variance

v <- c()
for (i in 1:1000) {
    v <- c(v, var(s[[i]]))
}

hist(v)
```

```r
#statistics = standard deviation

standard <- c()
for (i in 1:1000){
    standard <- c(standard, sd(s[[i]]))
}

hist(standard)
```

```r
#statistics = median

med <- c()
for (i in 1:1000){
    med <- c(med, median(s[[i]]))
}

hist(med)
```

```r
#statistcs = sum

su <- c()
for (i in 1:1000) {
    su <- c(su, sum(s[[i]]))
}

hist(su)
```

```r
#statistics = min

m <- c()
for (i in 1:1000){
    m <- c(m, min(s[[i]]))
}

hist(m)
```
