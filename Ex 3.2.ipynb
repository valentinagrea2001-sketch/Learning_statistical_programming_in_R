```r
#SAMPLING DISTRIBUTION OF STATISTICS

#If statistics= sample mean
#->-> by the CLT, the sampling distribution is approximatively normal

#Try for other statistics if the CLT holds:
#sample proportion
#sample variance
#sample median
#sample standard deviation
```

```r
#We choose this time an exponential random experiment
#a large population of N=10000, lambda=5

exp_experiment <- rexp(10000,5)

s <- list()
for (i in 1:200) {
    s[[i]] <- sample(exp_experiment, 500)   #sample sets n=500
}
```

```r
#statistics= proportion

vec <- lapply(s)  #vectorize s
table(vec)
```
