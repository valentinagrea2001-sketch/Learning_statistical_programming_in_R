```r
tab <- read.table("https://stat4ds.rwth-aachen.de/data/CovidMasks.dat", header=TRUE)
tab

#for each state, compare covid19 symptoms with mask use

#Two variables: Percent_Symptoms and Percent_Mask
```

```r
summary(tab$Percent_Symptoms)

sd(tab$Percent_Symptoms)

summary(tab$Percent_Mask)

sd(tab$Percent_Mask)

boxplot(tab$Percent_Mask, tab$Percent_Symptoms, xlab="Percent Mask", ylab="Percent Symptoms", main="COVID 19 symptoms vs mask use")

#quick comparison of the two variables
```

```r
#TO COMPARE TWO VARIABLES, USE A SCATTER PLOT
#is there a correlation between the two variables? i.e. do states with more symptoms have more mask use?

plot(tab$Percent_Mask, tab$Percent_Symptoms, xlab="Percent Mask", ylab="Percent Symptoms", main="COVID 19 symptoms vs mask use")

#explanatory variable on the x-axis: Percent_Mask
#response variable on the y-axis: Percent_Symptoms
```

```r
cov(tab$Percent_Mask, tab$Percent_Symptoms)   #not that useful

cor(tab$Percent_Mask, tab$Percent_Symptoms)   #normalised

#correlation coefficient <<0 : strong correlation and negative direction
```

```r
plot(tab$Percent_Mask, tab$Percent_Symptoms, xlab="Percent Mask", ylab="Percent Symptoms", main="COVID 19 symptoms vs mask use")
abline(lm(tab$Percent_Symptoms ~ tab$Percent_Mask))   #add regression line to the scatter plot
```
