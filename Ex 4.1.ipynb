```r
#STANDARD NORMAL DISTRIBUTION N(mu, sigma^2)

x <- seq(-20,20, by=0.1)
plot(x,dnorm(x, 0, 1), type="l", main="pdf of standard normal distribution", xlab="x", ylab="pdf")
```

```r
#compare gaussian distributions around the mean=0

plot(x,dnorm(x, 0, 1), type="l", main="pdf of standard normal distribution", xlab="x", ylab="pdf", col="red", lwd=3)
lines(x, dnorm(x, 0, 2), col="green", lwd=3)
lines(x, dnorm(x, 0, 4), col="yellow", lwd=3)
lines(x, dnorm(x, 0, 10), col="blue", lwd=3)

legend("topright", legend=c("N(0,1)", "N(0,2)", "N(0,4)", "N(0,10)"), col=c("red", "green", "yellow", "blue"), lty = 1, lwd=2)

#lty=1 for continuous line
#lty=2 for traits line

#lwd : how thick are the lines
```

```r
#CHI-SQUARED DISTRIBUTION >= 0
#d degrees of freedom
#more d increases, more the distribution approximates a normal distribution and the mean is skewed to the right

#for d>=30 degrees of freedom, there is already a good approximation to the normal distribution

x1 <- seq(0, 60, by=0.1)

plot(x1, dchisq(x1, 1), type="l", xlab="x", ylab="pdf", main="pdf of chi squared distribution", lwd="2")
lines(x1, dchisq(x1, 2), col="orange", lwd="2")
lines(x1, dchisq(x1, 4), col="green", lwd="2")
lines(x1, dchisq(x1, 10), col="blue", lwd="2")
lines(x1, dchisq(x1, 30), col="red", lwd="2")
```
