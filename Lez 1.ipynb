```r
#ANALYZING TABULAR DATA FROM A WEBSITE!!

carbon <- read.table("https://stat4ds.rwth-aachen.de/data/Carbon.dat", header=TRUE)
carbon

#header=true to keep the variables names in the file
```

```r
head(carbon,3) #display the first 3 rows of the data frame

tail(carbon,3) #display the last 3 rows of the data frame
```

```r
#EXTRACT DATA FROM THE TABLE
#notice that you can use the $ operator to refer to a specific variable in the data frame

max(carbon$CO2)     #find the maximum value of the CO2 variable

min(carbon$CO2)     #find the minimum value of the CO2 variable

mean(carbon$CO2)    #find the mean value of the CO2 variable

sd(carbon$CO2)      #find the standard deviation of the CO2 variable

median(carbon$CO2)  #find the median value of the CO2 variable, e.g. the middle value

Mode_compute <- function(x) {
    dup <- x                         #copy the values of CO2
    for (i in 1:length(x)) {         #compare to track the positions where the values are duplicated
        for (j in 1:length(x)) {
            if (x[i] == x[j]) {
                dup[i] <- dup[i] + 1
            }
        }
    }
    m = max(dup)                #find the maximum occurency
    return(x[which(dup == m)])  #return the value of CO2 that has the maximum occurency
}

Mode_compute(carbon$CO2)  #find the mode value of the CO2 variable, e.g. the most frequent value
```

```r
#FIND FREQUENCIES OF THE VALUES

#VALUES ARE CONTINOUS AND MUST BE GROUPED INTO INTERVALS 

#knowing that min-max is a range of [2,10),
#consider categories/intervals on x axes [2,3),...[9,10)
#and count the number of values that fall into each interval

#table(cut(..)) WORK TOGETHER TO FIND FREQUENCIES

breaks <- seq(2, 10, by = 1)    #2,3,..,10
c <- cut(carbon$CO2, breaks)   #maps continuous data into categories, e.g. [2,3),[3,4),..,[9,10)
freq <- table(c)              #counts occurrencies in each category

freq

```

```r
#FIND FREQUENCIES DISTRIBUTION

breaks <- seq(2, 10, by=1.0)
freq <- table(cut(carbon$CO2, breaks))
cbind(freq,freq/sum(freq))    #create a table with frequencies and distributions

```

```r
#FIND HISTOGRAM OF CO2

hist(carbon$CO2, main="Histogram of CO2", xlab="CO2", ylab="Frequency", col="lightblue", border="black")

hist(carbon$CO2, breaks=seq(2, 10, by=2.0), main="Histogram of CO2", xlab="CO2", ylab="Frequency", col="lightblue", border="black")

hist(carbon$CO2, breaks=seq(2, 10, by=0.5), main="Histogram of CO2", xlab="CO2", ylab="Frequency", col="lightblue", border="black")
```

```r
#FIND SMOOTH APPROXIMATION OF THE HISTOGRAM

plot(density(carbon$CO2), main="Density of CO2", xlab="CO2", ylab="Density", col="blue", lwd=2)
```

```r
#SUMMARY STATISTICS
#i.e. mean and mean, lowe and upper quartiles, min and max values

summary(carbon$CO2)
```

```r
#BOX-PLOT

boxplot(carbon$CO2, main="Boxplot of CO2", ylab="CO2", col="lightblue", border="black")
```
