```r
#SAMPLING A RANDOM SUBSET

sample(1:20, 5)                  #sample 5 numbers from a range of 1 to 20 without repetitions
                                 #default mode is replace=FALSE

sample(1:20, 5, replace=TRUE)    #sample 5 numbers from a range of 1 to 20 with repetitions
```

```r
#ANALYZING TABULAR DATA FROM A WEBSITE!!

carbon <- read.table("https://stat4ds.rwth-aachen.de/data/Carbon.dat", header=TRUE)
carbon

#header=true to keep the variables names in the file
```

```r
head(carbon,3) #display the first 3 rows of the data frame

tail(carbon,3) #display the last 3 rows of the data frame
```

```r
#EXTRACT DATA FROM THE TABLE
#notice that you can use the $ operator to refer to a specific variable in the data frame

max(carbon$CO2)     #find the maximum value of the CO2 variable

min(carbon$CO2)     #find the minimum value of the CO2 variable

mean(carbon$CO2)    #find the mean value of the CO2 variable

median(carbon$CO2)  #find the median value of the CO2 variable, e.g. the middle value

Mode_compute <- function(x) {
    ux <- unique(x)                           # Get all unique values from x
    
    match_result <- match(x, ux)              # For each element in x, find its position in ux
                                               # Example: if x = c(1,2,1,3,2,2), ux = c(1,2,3)
                                               # match_result = c(1, 2, 1, 3, 2, 2)
    
    freq_table <- tabulate(match_result)      # Count occurrences of each position
                                               # Returns: c(2, 3, 1) meaning: 1 appears 2x, 2 appears 3x, 3 appears 1x
    
    max_position <- which.max(freq_table)     # Find which position has the highest count
                                               # Returns: 2 (because 3 is the max frequency)
    
    mode_value <- ux[max_position]            # Get the actual value at that position
                                               # Returns: ux[2] = 2 (the most frequent value)
    
    return(mode_value)
}

Mode_compute(carbon$CO2)  #find the mode value of the CO2 variable, e.g. the most frequent value
    
```
