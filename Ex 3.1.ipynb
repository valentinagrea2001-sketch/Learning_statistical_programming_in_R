```r
#TOPIC: SAMPLING DISTRIBUTIONS

#Central Limit Theorem
#The sampling distribution of samples means approximates the normal distribution,
#considering large sample sets of same size
```

```r
#Run a Poisson Experiment with parameter lambda=5
#over a large population of N=10000 individuals

poi_exp <- rpois(10000, 5)

#Choose sample size n=100

s <- list()            #list of 1000 samples sets
for (i in 1:1000) {
    s[[i]] <- sample(poi_exp, 100)
}

#Compute the mean for each sample

m <- c()                #vector of entries the mean
for (i in 1:1000) {
    m <- c(m, mean(s[[i]]))
}

#Mean of m should be around that of actual experiment, so 5

mean(m)

#Variance of m should be around 5/n=5/100=0.05

var(m)

#Standard deviation of m

sd(m)

#Plot the histogram of sampling means

hist(m, main="frequency of mean samples for a poisson experiment", xlab="means of samples", ylab="frequency")
```

```r
#Run a Poisson Experiment with parameter lambda=5
#over a large population of N=10000 individuals

poi_exp <- rpois(10000, 5)

#Choose sample size n=100

s <- list()            #list of 1000 samples sets
for (i in 1:1000) {
    s[[i]] <- sample(poi_exp, 100)
}

#Compute the mean for each sample

m <- c()                #vector of entries the mean
for (i in 1:1000) {
    m <- c(m, mean(s[[i]]))
}

#Mean of m should be around that of actual experiment, so 5

mean(m)

#Variance of m should be around 5/n=5/100=0.05

var(m)

#Standard deviation of m

sd(m)

#Plot the histogram of sampling means

h <- hist(m, plot=FALSE)
names(h)
h$counts <- h$counts/sum(h$counts)

plot(h, main="sampling distribution of mean samples for a poisson experiment", xlab="means of samples", ylab="probability", ylim=c(0,1))
axis(2, at=seq(0,1,by=0.05))
```

```r
#Poisson experiment with a small finite population of N=50 indoviduals
#therefore, the size of sample sets must be n < N/10=5 by the 10% rule
#We choose n=4

poi_exp <- rpois(50, 3)   #lambda=3

#Random sampling of subsets of size 4 from the experiment

s <- list()
for (i in 1:20) {
    s[[i]] <- sample(poi_exp, 4)
}

#Compute the mean on each random sample set

m <- c()  #vector
for (i in 1:20) {
    m <- c(m, mean(s[[i]]))
}

#Plot the distribution of means

h <- hist(m, plot=FALSE)
h$counts <- h$counts/sum(h$counts)

plot(h)
axis(2, at=seq(0,1,by=0.05))
```
